```python
from groq import Groq
import base64
import os

# Initialize Groq client ONCE
client = Groq(api_key=os.environ.get("GROQ_API_KEY"))

def ocr_image_with_groq(image_path):
    """
    Takes image path
    Returns extracted text from image using Groq Vision
    """

    # Encode image
    with open(image_path, "rb") as image_file:
        base64_image = base64.b64encode(image_file.read()).decode("utf-8")

    # Call Groq Vision model
    chat_completion = client.chat.completions.create(
        messages=[
            {
                "role": "user",
                "content": [
                    {"type": "text", "text": "Extract all readable text from this medical report image."},
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": f"data:image/jpeg;base64,{base64_image}",
                        },
                    },
                ],
            }
        ],
        model="meta-llama/llama-4-scout-17b-16e-instruct",
    )

    extracted_text = chat_completion.choices[0].message.content

    # SAFETY CHECK (VERY IMPORTANT)
    if not extracted_text or len(extracted_text.strip()) == 0:
        raise ValueError(f"OCR failed for image: {image_path}")

    return extracted_text
```
