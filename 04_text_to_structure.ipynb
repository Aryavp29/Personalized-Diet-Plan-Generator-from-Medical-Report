```python
import os
import re
import pandas as pd
```

```python
TEXT_DIR = "data/extracted_text"

def read_text(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        return f.read()
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"\s+", " ", text)      # remove extra spaces
    return text
```

```python
def extract_value(text, keywords):
    for keyword in keywords:
        pattern = rf"{keyword}[^0-9]*([0-9]+\.?[0-9]*)"
        match = re.search(pattern, text)
        if match:
            return float(match.group(1))
    return None
```

```python
def extract_medical_data(text):
    return {
        # CBC
        "hemoglobin": extract_value(text, ["hemoglobin"]),
        "rbc_count": extract_value(text, ["rbc count"]),
        "wbc_count": extract_value(text, ["wbc count", "total wbc"]),
        "platelet_count": extract_value(text, ["platelet count"]),
        "neutrophils_pct": extract_value(text, ["neutrophils"]),
        "lymphocytes_pct": extract_value(text, ["lymphocytes"]),

        # Sugar / Diabetes
        "fasting_glucose": extract_value(text, ["fasting blood sugar", "fasting glucose"]),
        "pp_glucose": extract_value(text, ["blood sugar pp", "post prandial"]),
        "hba1c": extract_value(text, ["hba1c", "glycosylated hemoglobin"]),
        "mean_blood_glucose": extract_value(text, ["mean blood glucose"]),

        # Lipid profile
        "total_cholesterol": extract_value(text, ["cholesterol total", "total cholesterol"]),
        "triglycerides": extract_value(text, ["triglyceride"]),
        "hdl": extract_value(text, ["hdl cholesterol", "hdl"]),
        "ldl": extract_value(text, ["ldl cholesterol", "direct ldl"]),
        "vldl": extract_value(text, ["vldl"]),

        # Kidney
        "urea": extract_value(text, ["urea"]),
        "creatinine": extract_value(text, ["creatinine"]),
        "uric_acid": extract_value(text, ["uric acid"]),

        # Thyroid
        "t3": extract_value(text, ["t3", "triiodothyronine"]),
        "t4": extract_value(text, ["t4", "thyroxine"]),
        "tsh": extract_value(text, ["tsh", "thyroid stimulating hormone"]),

        # Vitamins
        "vitamin_d": extract_value(text, ["25oh vitamin d", "vitamin d"]),
        "vitamin_b12": extract_value(text, ["vitamin b12"]),
    }
```

```python
records = []

for file in os.listdir(TEXT_DIR):
    file_path = os.path.join(TEXT_DIR, file)
    
    if not file.endswith(".txt"):
        continue

    raw_text = read_text(file_path)
    cleaned_text = clean_text(raw_text)
    
    data = extract_medical_data(cleaned_text)
    data["report_id"] = file.replace(".txt", "")
    
    records.append(data)

records
```

```python
df = pd.DataFrame(records)
df.to_csv("data/final_dataset.csv", index=False)

df
```
